# NumPy 广播机制实验

## 一、实验目的

理解 NumPy 广播机制的原理，并通过 Python 程序验证不同形状数组之间的运算。

## 二、广播机制原理

广播是 NumPy 在数组形状不完全相同时，使数组能够进行逐元素运算的机制。NumPy 会从最后一个维度开始比较两个数组的形状：

- 两个维度大小相同，可以运算；
- 其中一个维度大小为 `1`，可以扩展后运算；
- 标量可以与任意形状的数组运算；
- 如果对应维度既不相等，也没有一个为 `1`，则无法广播。

广播通常只是改变数组的逻辑视图，不会简单地复制大量数据，因此能够减少代码量并提高计算效率。

## 三、实验验证

下面分别验证标量广播、同形数组运算、二维数组广播，以及形状不兼容时的错误。

### 标量广播


```python
import numpy as np

numbers = np.array([1, 2, 3, 4])
result = numbers + 10

print("原数组：", numbers)
print("加上标量 10：", result)
```

### 同形数组运算


```python
import numpy as np

first = np.array([1, 2, 3])
second = np.array([10, 20, 30])
result = first + second

print("第一个数组：", first)
print("第二个数组：", second)
print("逐元素相加：", result)
```

### 二维数组广播


```python
import numpy as np

column = np.array([[1], [2], [3]])
row = np.array([[10, 20, 30, 40]])
result = column + row

print("列数组形状：", column.shape)
print("行数组形状：", row.shape)
print("结果形状：", result.shape)
print("广播结果：\n", result)
```

### 不兼容形状


```python
import numpy as np

left = np.ones((2, 3))
right = np.ones((2, 2))

try:
    result = left + right
except ValueError as error:
    print("无法广播：", type(error).__name__)
    print("原因：两个数组的对应维度不满足匹配规则")
```

## 四、实验结论

实验验证了 NumPy 会按照广播规则自动扩展兼容的数组形状。标量可以和任意形状的数组运算，形状为 `(3, 1)` 和 `(1, 4)` 的数组可以广播为 `(3, 4)`；而 `(2, 3)` 和 `(2, 2)` 的数组由于对应维度不匹配，无法进行广播。
