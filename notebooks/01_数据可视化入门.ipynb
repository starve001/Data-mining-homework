# numpy数组的广播机制是：
1. 数组维度必须一致
2. 数组中元素个数必须一致或者至少有一个数组的元素个数等于1
3. 数组中元素个数一致时，两个数组对应位置的元素必须满足broadcastable的条件
4. 如果两个数组对应位置的元素满足broadcastable的条件，则两个数组的对应位置的元素进行broadcast
5. 如果两个数组对应位置的元素不满足broadcastable的条件，则两个数组的对应位置的元素进行广播

```python
import numpy as np
import matplotlib.pyplot as plt
time =['20200401','20200402','20200403','20200404','20200405']
china = [93,78,73,55,75]
math = [78,45,34,56,28]
plt.figure(figsize=(5,4),dpi=100)
plt.plot(time,china)
plt.plot(time,math)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 固定种子数：必须放在所有随机数生成之前，保证实验结果可重复
np.random.seed(19680801)

# 产生50对服从标准正态分布的样本点
nbPoints = 50
x = np.random.standard_normal(nbPoints)
y = np.random.standard_normal(nbPoints)

# 生成随机颜色映射值和散点面积
colors = np.random.rand(nbPoints)
area = (30 * np.random.rand(nbPoints)) ** 2

# 绘制散点图：s控制点大小，c控制点颜色，alpha控制透明度
plt.scatter(x, y, s=area, c=colors, alpha=0.5)
plt.show()
```

```python
import pandas as pd

# 读取2015、2016、2017三份独立csv文件，分别存入DataFrame
# 2015和2017文件使用逗号分隔，2016文件使用分号分隔
df2015 = pd.read_csv('../data/2015_data.csv', sep=',')
df2016 = pd.read_csv('../data/2016_data.csv', sep=';')
df2017 = pd.read_csv('../data/2017_data.csv', sep=',')
# 预览2017数据集前5行
df2017.head()
```

```python
df2015.head()
```

```python
df2016.head()
```

```python
df=pd.concat([df2015,df2016,df2017],axis=0)

df.head()
```

```python
df.index=[i for i in range(df.shape[0])]
df.columns=['area', 'year', 'total_value', 'one_value', 'two_value',
    'three_value','zero_value', 'inout', 'population', 'pay',
    'student', 'hospitalt', 'realty']
df.head(15)
# 原列名与重置后的列名对应关系:
# area                 地区
# year                 年份
# total_value        国内生产总值
# one_value          第一产业增加值
# two_value          第二产业增加值
# three_value        第三产业增加值
# zero_value         社会商品零售总额
# inout              货物进出口总额
# population         年末总人口
# pay                在岗职工平均工资
# student            普通高等学校在校学生数
# hospitalt          医院、卫生院数
# realty             房地产开发投资额
```

```python
df.isnull().sum()
```

```python
df['realty']=df['realty'].fillna(value=df['realty'].mean())
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt

#设置图像风格:
plt.style.use('ggplot')

#设置字体格式及负值显示
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

#设置图像标题
plot_title = ['图1: 国内生产总值','图2: 第一产业增加值','图3: 第二产业增加值',
              '图4: 第三产业增加值','图5: 社会商品零售总额','图6: 货物进出口总额',
              '图7: 在岗职工平均工资','图8: 普通高等学校在校学生数','图9: 医院、卫生院数',
              '图10: 房地产开发投资额']
```

```python
#获取各年
dt2015 = df[df['year'] == 2015]
dt2016 = df[df['year'] == 2016]
dt2017 = df[df['year'] == 2017]
#定义列表
dataArr = ['area','total_value', 'one_value', 'two_value', 'three_value','zero_value', 'inout', 'pay','student', 'hospitalt', 'realty']
```

```python
data = dataArr
#设置步画
fig = plt.figure(figsize=(50,20),dpi=100  )
#4-2选择绘制线型图
plt.plot(dt2015[data[0]],dt2015[data[1]],marker='o',label='2015')
plt.plot(dt2016[data[0]],dt2016[data[1]],marker='o',label='2016')
plt.plot(dt2017[data[0]],dt2017[data[1]],marker='o',label='2017')
#标题
plt.title(plot_title[0])
plt.legend(loc='best')
plt.show()  
```

```python
def plotType(data):
    fig, axes = plt.subplots(5, 2, figsize=(18, 25))
    axes = axes.ravel()

    for i, ax in enumerate(axes):
        value_column = data[i + 1]
        ax.plot(dt2015[data[0]], dt2015[value_column], marker='o', label='2015')
        ax.plot(dt2016[data[0]], dt2016[value_column], marker='o', label='2016')
        ax.plot(dt2017[data[0]], dt2017[value_column], marker='o', label='2017')
        ax.set_title(plot_title[i])
        ax.set_xticks(range(len(dt2015[data[0]])))
        ax.set_xticklabels(dt2015[data[0]], rotation=90)
        ax.legend(loc='best')

    fig.tight_layout()
    plt.show()


plotType(dataArr)
```
