```python
# sKLEARN算法
from sklearn.linear_model import LinearRegression   # 导入线性回归
#2.内置数据集
from sklearn import datasets

#3.自定义数据集
from sklearn.datasets import load_iris
iris = load_iris()



```

```python
#1.加载鸢尾花数据集
diabetes = datasets.load_diabetes()
```

```python
#2.选择模型
line_model = LinearRegression()#线性回归模型
```

```python
#3.训练模型
line_model.fit(diabetes.data,diabetes['target'])
#line_model.score(diabetes.data,diabetes.target)
```

```python
#4.预测,先取前10行数据
line_model.predict(diabetes.data[:10])
#另一种写法
line_model.predict(diabetes.data[:10,:])
```

```python
diabetes.target[:10]
#属性10行
```

```python
#5.评估模型
line_model.score(diabetes.data,diabetes.target)
```

### 方法有问题？应该采用多元回归模型吗？

### 可能不是线性关系，非线性模拟

### 非线性特点：
- 输入数据特征维度高
- 输入数据特征分布不均匀
- 输入数据特征之间存在依赖关系
- 输入数据特征之间存在非线性关系
- 输入数据特征之间存在多态

```python
#原数据进行多项式展开
#数据转换，即数据的预处理：processing
from sklearn.preprocessing import PolynomialFeatures
```

```python
# 1.实例化
pf = PolynomialFeatures(degree=10)
```

```python
#2.训练模型
pf.fit(diabetes.data)
data_new = pf.transform(diabetes.data)
data_new.shape
#66是
data_new
```

```python
line_model_new = LinearRegression()
line_model_new.fit(data_new,diabetes.target)
```

```python
line_model_new.score(data_new,diabetes.target)
```

## 实验总结

本节学习了 `scikit-learn` 的基本使用流程：首先加载内置数据集，了解数据集中的特征和标签；然后以糖尿病数据集为例，实例化 `LinearRegression` 线性回归模型，使用 `fit()` 完成训练，通过 `predict()` 进行预测，并使用 `score()` 对模型效果进行评估。

实验还尝试了多项式特征方法。由于原始数据的特征之间可能存在非线性关系，使用 `PolynomialFeatures(degree=10)` 对数据进行多项式展开，再将转换后的数据输入线性回归模型进行训练和评分。由此可见，线性回归也可以通过特征变换拟合一定程度的非线性关系，但多项式次数过高会显著增加特征数量，可能带来计算量增大和过拟合问题，因此实际应用中需要结合模型评分、数据规模和验证集结果选择合适的参数。
