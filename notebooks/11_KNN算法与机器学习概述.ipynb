## 导入KNN算法


什么是KNN算法
KNN算法（K-Nearest Neighbors）是一种基于“最近邻”的分类算法。KNN算法的原理是：给定一个样本，通过计算该样本与训练数据集中其他样本之间的距离，然后找出距离最近的K个样本，将这些样本的类别进行投票，最终得到该样本的类别。KNN算法的优点是简单、高效，缺点是计算距离的复杂度较高，且K的选择对结果影响较大。
KNN算法的实现步骤如下：

1. 数据预处理：首先对数据进行预处理，包括数据清洗、数据转换和数据归一化。

2. 划分训练集和测试集：将数据集划分为训练集和测试集，通常80%的数据用于训练，20%的数据用于测试。

3. 计算距离：根据距离度量方法计算训练数据集中的每个样本与测试数据集中的每个样本之间的距离。常用的距离度量方法有欧氏距离、曼哈顿距离和切
氏距离。

# K近邻算法


k一般是奇数，因为当k为偶数时，存在平局的情况，无法确定分类结果。

````python
#导入所需要的包
import numpy as np
import matplotlib.pyplot as plt
from math import sqrt
from collections import Counter

def distance(k, X_train, Y_train, x):
    #保证K有效
   # assert 1 <= k <= X_train.shape[0], "K must be valid"
    #X_train的样本数量必须等于y_train的样本数量
    #assert X_train.shape[0] == Y_train.shape[0], "the size of X_train must equal to the size of y_train"
    #x的特征号必须等于X_train
    #assert X_train.shape[1] == x.shape[0], "the feature number of x must be equal to X_train"
    
    #计算欧氏距离，for循环来遍历所有样本
    distance = [sqrt(np.sum((x_train - x)**2)) for x_train in X_train]
    #返回距离值从小到大排序后的索引值的数组
    nearest = np.argsort(distance)
    #获取距离最小的前k个样本的标签
    topk_y = [Y_train[i] for i in nearest[:k]]
    #统计前k个样本的标签类别以及对应的频数
    votes = Counter(topk_y)
    #返回频数最多的类别
    return votes.most_common(1)[0][0]
````

````python
#使用numpy生成8个点
X_train = np.array([[1.0, 3.5],
                    [2.0, 7],
                    [3.0, 10.5],
                    [4.0, 14],
                    [5, 25],
                    [6, 30],
                    [7, 35],
                    [8, 40]])

#使用numpy生成8个点对应的类别
Y_train = np.array([0, 0, 0, 0, 1, 1, 1, 1])

#使用numpy生成待分类样本点
x = np.array([8, 21])

#调用distance函数并传入参数
label = distance(3, X_train, Y_train, x)

#显示待测样本点的分类结果
print(label)
````

## 原理：KNN中0和1怎么判断（核心：**邻居投票，少数服从多数**）
我们的数据：
- 类别 **0**：前4个点：`(1,3.5)、(2,7)、(3,10.5)、(4,14)`
- 类别 **1**：后4个点：`(5,25)、(6,30)、(7,35)、(8,40)`
- 待测点：`x = (8, 21)`，k=3，找距离最近**3个邻居**

## 第一步：计算待测点(8,21)到全部8个点的欧氏距离
欧氏距离公式：
$$d=\sqrt{(x_2-x_1)^2+(y_2-y_1)^2}$$

|训练点|类别|距离到(8,21)|
|---|---|---|
|(5,25)|**1**|5.00 ✔|
|(4,14)|**0**|8.06 ✔|
|(6,30)|**1**|10.30 ✔|
|(3,10.5)|0|13.82|
|(7,35)|1|14.14|
|(2,7)|0|15.65|
|(8,40)|1|19.00|
|(1,3.5)|0|19.34|

距离从小到大排序，**取前3个最近的**：
1. (5,25) →类别`1`
2. (4,14) →类别`0`
3. (6,30) →类别`1`

> 得到3个邻居标签列表：`topk_y = [1, 0, 1]`

## 第二步 Counter 统计投票次数
```python
votes = Counter([1,0,1])
```
统计结果：
- `1` 出现 **2次**
- `0` 出现 **1次**

## 第三步 most_common(1)[0][0]选出票数最多
`votes.most_common(1)` 返回：`[(1, 2)]`
- `(1,2)`：元组，第一个是类别，第二个是票数
- `[0]`取这个元组 →`(1,2)`
- `[0]`取出类别 →`1`

✅ 所以最终返回结果是 **1**

---

### 例1：k=1（只看最近1个）

### 例2：k=5（取最近5个）

统计：1出现3次，0出现2次 →输出`1`

> 手写这个简易版本没有处理平票！平票会取Counter内部顺序，这是手写KNN的缺陷。

---
# 通俗大白话
> 0、1只是给两类样本打的**标签代号**，本身没有数学含义，不是数字大小。
> KNN**不计算待测点等于0还是等于1**，不做大小比较！
> 逻辑：**看周围k个已经分好类的邻居，邻居里面0多就判0，1多就判1。**

````python
import numpy as np
import matplotlib.pyplot as plt
from math import sqrt
from collections import Counter

def knn_predict(X_train, Y_train, x, k):
    distances = [((x_train - x) ** 2).sum() ** 0.5 for x_train in X_train]
    nearest = sorted(range(len(distances)), key=lambda i: distances[i])[:k]
    return Counter(Y_train[i] for i in nearest).most_common(1)[0][0]
````

````python
import numpy as np
from collections import Counter

def distance(k, X_train, Y_train, x):
    d = np.linalg.norm(X_train - x, axis=1) 
      # 计算每个点到 x 的距离
    # 对距离进行排序
    nearest = np.argsort(d)[:k]               # 取最近的 k 个
    return Counter(Y_train[nearest]).most_common(1)[0][0]

X_train = np.array([[1.0, 3.5],
                    [2.0, 7],
                    [3.0, 10.5],
                    [4.0, 14],
                    [5, 25],
                    [6, 30],
                    [7, 35],
                    [8, 40]])

Y_train = np.array([0, 0, 0, 0, 1, 1, 1, 1])
x = np.array([8, 21])

label = distance(3, X_train, Y_train, x)
print(label)
````
