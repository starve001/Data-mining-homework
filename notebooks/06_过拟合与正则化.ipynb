# 过拟合


## 模型误差与过拟合

### 模型误差的组成

模型的总误差通常可以理解为由以下三部分组成：

- **偏差（Bias）**：模型预测值与真实值之间的系统性误差。模型过于简单，无法充分学习数据中的规律时，偏差通常较大，这种情况容易出现欠拟合。
- **方差（Variance）**：模型预测结果之间的离散程度，反映模型对训练数据变化的敏感程度。模型过于复杂时，容易过度学习训练数据中的细节和噪声，导致方差较大。
- **数据本身的误差**：数据采集、测量或不可避免的随机因素造成的误差，通常无法完全消除。

### 什么是过拟合

过拟合是指模型在训练集上的表现非常好，但在新的、未见过的数据上的表现较差。其主要原因是模型过于复杂，除了学习数据的真实规律外，还学习了训练数据中的噪声和偶然性，因此泛化能力不足。

## 偏差与方差的权衡

偏差和方差通常存在矛盾关系：

- 降低偏差，往往需要提高模型复杂度，但可能导致方差增大；
- 降低方差，往往需要限制模型复杂度，但可能导致偏差增大。

因此，不能只追求更复杂的模型或更低的训练误差，而应该在偏差和方差之间寻找平衡，使模型在测试数据上的总误差尽可能小。通常随着模型复杂度增加，偏差会下降，方差会升高，而总误差可能呈现先下降后上升的趋势。

### 减少过拟合的方法

- 选择合适的模型复杂度和多项式阶数；
- 增加训练数据量，提高数据的代表性；
- 使用正则化方法限制模型参数；
- 通过交叉验证选择合适的模型和超参数；
- 删除无关特征或进行特征选择；
- 同时关注训练集、验证集和测试集表现。

## 正则化与 Ridge 回归

### 为什么需要正则化

普通线性回归通常通过最小化均方误差（MSE）来拟合模型：

$$
MSE = \sum_i \left(y_i - f(x_i)\right)^2
$$

当模型特征较多、特征之间存在较强相关性，或模型复杂度较高时，线性回归可能出现过拟合。为限制参数大小、降低过拟合风险，可以在损失函数中加入正则化项。

### Ridge 回归

Ridge 回归在 MSE 的基础上加入 L2 范数惩罚项：

$$
L_{Ridge} = MSE + \alpha \sum_i w_i^2
$$

其中，$w_i$ 表示模型的回归系数，$\alpha$ 是正则化强度。Ridge 回归会使回归系数整体变小并接近于 0，但通常不会将系数完全变为 0，因此模型仍会保留较多特征。

## Lasso 回归

Lasso 回归在 MSE 的基础上加入 L1 范数惩罚项：

$$
L_{Lasso} = MSE + \alpha \sum_i |w_i|
$$

L1 正则化会使一部分回归系数直接变为 0，因此 Lasso 回归不仅可以缓解过拟合，还可以进行特征选择，得到更加稀疏、易解释的模型。

## Ridge 与 Lasso 的区别

| 对比方面 | Ridge 回归 | Lasso 回归 |
| --- | --- | --- |
| 正则化类型 | L2 正则化 | L1 正则化 |
| 惩罚项 | $\sum_i w_i^2$ | $\sum_i |w_i|$ |
| 对系数的影响 | 使系数接近 0 | 可使部分系数变为 0 |
| 特征选择 | 通常不能直接选择特征 | 可以进行特征选择 |
| 适用特点 | 特征之间存在相关性，希望保留更多特征 | 特征较多，希望得到稀疏模型 |

## `alpha` 参数的作用

`alpha` 用于控制正则化强度：

- `alpha` 越大，系数收缩越明显，模型复杂度越低，抗过拟合能力通常越强；
- `alpha` 越小，模型越接近普通线性回归，可能保留更多数据特征，但过拟合风险也可能增加。

实际应用中应通过验证集或交叉验证选择合适的 `alpha`，在模型拟合能力和泛化能力之间取得平衡。
