```python
from sklearn.linear_model import LogisticRegression
from sklearn import datasets
from sklearn.datasets import load_breast_cancer
#用于数据拆分
from sklearn.model_selection import train_test_split
```

```python
data_bc = datasets.load_breast_cancer()
```

```python
data_bc
```

```python
#stratify:防止训练集与测试集类别偏差过大
#返回：训练集属性，测试集属性，训练集标签，测试集标签
X_train,X_test,y_train,y_test=train_test_split(data_bc.data,data_bc.target,test_size=0.25,stratify=data_bc.target)
```

```python
from sklearn.preprocessing import StandardScaler
ss = StandardScaler()
#分别对训练集和测试集进行标准化
X_train_new = ss.fit_transform(X_train)
#对测试集进行标准化
X_test_new = ss.transform(X_test)
```

```python
#2.选择模型
logi_model = LogisticRegression(max_iter=500)#逻辑回归模型

```

```python
#3.训练模型
logi_model.fit(X_train_new,y_train)
#logi_model.score(data_bc.data,data_bc.target)
```

```python
y_predict = logi_model.predict(X_test_new)
logi_model.score(X_test_new,y_test)
```

```python
y_test
```

```python
y_predict
```

## 各个类别概率预测

```python
y_proba = logi_model.predict_proba(X_test_new)
```

```python
#该组数据只有两个类别
y_proba
```

```python
#取出预测为1-良性概率
result_1 = y_proba[:,1]#类别1的概率
result_2 = y_proba[:,0]#类别0的概率
```

```python
#拆分出1为良性，0恶性的概率
#实际为1良性，预测为1对应的概率是多少
predict_good = result_1[y_test == 1] 
#实际为0恶性，预测为0对应的概率是多少
predict_bad = result_1[y_test == 0] 
```

```python
#可视化表示
import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif']= ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False
```

```python
plt.figure(figsize=(10,5))
plt.hist(predict_bad,label='恶性，标签为1',color='red')
plt.hist(predict_good,label='良性，标签为0',color='green')
plt.axvline(x=0.5,color='black',linestyle='--',linewidth='2',label='决策边界x=0.5')
plt.legend()
```

### 参数调优


```python
from sklearn.model_selection import GridSearchCV
```

```python
lg_gs = LogisticRegression(solver='liblinear')
```

## 需要调整哪些参数，每个参数分别使用哪些值
## 字典的key一定是对应模型参数
## key对应的value是一个列表，列表中是该参数的所有取值，一定是参数可用的值

```python
# 用法
# 参数
# 1. estimator: 需要调参的模型实例
# 2. param_grid: 需要调整的参数字典，key为需要调参的参数名，value为该参数的取值列表
# 3. cv: 交叉验证的折数，默认为5；每种参数训练评分次数
# 4. scoring: 模型评估指标，默认为None，使用estimator的默认评估指标
gs_estimator = GridSearchCV(lg_gs, param_grid={'C':[0.01,0.1,1,10,100]}, cv=5, scoring='accuracy')
```

```python
# 训练
gs_estimator.fit(X_train_new,y_train)
```

```python
gs_estimator.best_params_
```

```python
gs_estimator.best_estimator_
```

```python
gs_estimator.best_score_
```
