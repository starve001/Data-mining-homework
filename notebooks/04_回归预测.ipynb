# 多元线性回归代码思路详解
数据集：2个自变量 $x_1$(里程)\(w_1、w_2\)$x_2$(次数)\(x_1=100\(x_1=102，x_2=6\)x_2=4\)1个因变量$y$(运输时间)\(x_1=100\(x_1=102，x_2=6\)x_2=4\)公式：
$$y = w_1x_1 + w_2x_2 + b$$

```python
import numpy as np
from sklearn.linear_model import LinearRegression
```
1. **导入库**
- `numpy`：用来构造二维特征数组\(x_1=100\(x_1=102，x_2=6\)x_2=4\)sklearn要求输入特征必须是二维
- `LinearRegression`：sklearn内置多元线性回归类\(x_1=100\(x_1=102，x_2=6\)x_2=4\)一元\(w_1、w_2\)多元都用这个类

---
```python
X_train = np.array([
    [100,4],
    [50,3],
    [100,4],
    [100,2],
    [50,2],
    [80,2],
    [75,3],
    [65,4],
    [90,3],
    [90,2]
])
```
2. **构造特征X_train（自变量）**
> 多元：**每一行代表一条样本\(x_1=100\(x_1=102，x_2=6\)x_2=4\)每一列代表一个特征**
- 样本1：$x_1=100\(x_1=100\(x_1=102，x_2=6\)x_2=4\)x_2=4$ → `[100,4]`
- 一共10条样本\(x_1=100\(x_1=102，x_2=6\)x_2=4\)2列\(x_1=100\(x_1=102，x_2=6\)x_2=4\)shape为`(10,2)`
> ⚠️sklearn硬性规则：X必须二维\(x_1=100\(x_1=102，x_2=6\)x_2=4\)就算1个特征也要写成`[[x]]`\(x_1=100\(x_1=102，x_2=6\)x_2=4\)多个特征直接每行放多个值。

```python
y_train = np.array([9.3,4.8,8.9,6.5,4.2,6.2,7.4,6.0,7.6,6.1])
```
3. **构造标签y_train（因变量\(x_1=100\(x_1=102，x_2=6\)x_2=4\)输出结果）**
y是每条样本对应的真实运输时间\(x_1=100\(x_1=102，x_2=6\)x_2=4\)一维数组\(x_1=100\(x_1=102，x_2=6\)x_2=4\)顺序和X_train行一一对应。
第一行X`[100,4]`对应y第一个值`9.3`。

---
```python
multi_model = LinearRegression()
```
4. **实例化模型对象**
创建多元线性回归模型\(x_1=100\(x_1=102，x_2=6\)x_2=4\)此时 $w_1,w_2,b$ 都是空的\(x_1=100\(x_1=102，x_2=6\)x_2=4\)还没有学习数据。

```python
multi_model.fit(X_train, y_train)
```
5. **模型训练 `.fit(特征,标签)`**
核心步骤\(x_1=100\(x_1=102，x_2=6\)x_2=4\)输入全部样本X和真实y\(x_1=100\(x_1=102，x_2=6\)x_2=4\)模型内部最小二乘法计算出最优权重 $w_1\(w_1、w_2\)w_2$ 和截距$b$\(x_1=100\(x_1=102，x_2=6\)x_2=4\)**直接修改模型对象本身\(x_1=100\(x_1=102，x_2=6\)x_2=4\)没有返回值**。

训练完成后\(x_1=100\(x_1=102，x_2=6\)x_2=4\)模型内部得到方程：
$$\hat y = w_1\cdot x_1 + w_2\cdot x_2 +b$$

```python
X_test = np.array([[102,6]])
y_pred = multi_model.predict(X_test)
```
6. **预测 `.predict(待预测特征)`**
任务：里程102\(x_1=100\(x_1=102，x_2=6\)x_2=4\)运输6次。
`[[102,6]]`：新样本\(x_1=100\(x_1=102，x_2=6\)x_2=4\)同样二维格式。
`predict()`把$x_1=102,x_2=6$代入已经训练好的公式\(x_1=100\(x_1=102，x_2=6\)x_2=4\)算出预测运输时间\(x_1=100\(x_1=102，x_2=6\)x_2=4\)返回预测结果。

```python
print("系数coef：", multi_model.coef_)
print("截距intercept：", multi_model.intercept_)
print("预测运输时间：", y_pred[0])
```
7. **读取训练出来的参数**
- `coef_`：数组 $[w_1,w_2]$\(x_1=100\(x_1=102，x_2=6\)x_2=4\)两个自变量的权重\(x_1=100\(x_1=102，x_2=6\)x_2=4\)顺序和X的列顺序一致（第一列x1\(x_1=100\(x_1=102，x_2=6\)x_2=4\)第二列x2）
- `intercept_`：截距 $b$

---
## 关键对比：一元 vs 多元
- 一元：$y=w_1x_1+b$\(x_1=100\(x_1=102，x_2=6\)x_2=4\)X每行只有1个值 `[[1],[3]]`
- 多元（本题2个变量）：$y=w_1x_1+w_2x_2+b$\(x_1=100\(x_1=102，x_2=6\)x_2=4\)X每行写多个特征`[100,4]`

> `LinearRegression`不需要改任何其他代码\(x_1=100\(x_1=102，x_2=6\)x_2=4\)只要X的列数增加\(x_1=100\(x_1=102，x_2=6\)x_2=4\)自动切换多元回归。
