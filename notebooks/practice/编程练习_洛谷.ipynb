# B2004 对齐输出

## 题目描述

读入三个整数，按每个整数占 $8$ 个字符的宽度，右对齐输出它们，按照格式要求依次输出三个整数，之间以一个空格分开。

## 输入格式

只有一行，包含三个整数 $a,b,c$。整数之间以一个空格分开。

## 输出格式

只有一行，按照格式要求依次输出三个整数，之间以一个空格分开。

## 输入输出样例 #1

### 输入 #1

```
123456789 0 -1

```

### 输出 #1

```
123456789        0       -1

```

## 说明/提示

对于 $100 \%$ 的数据，$-2^{31} \le a, b, c < 2^{31}$。

````python
a = int(input())
b = int(input())
c = int(input())
print(f"{a:8}{b:8}{c:8}")   
````

# P1002 [NOIP 2002 普及组] 过河卒

## 题目描述

棋盘上 $A$ 点有一个过河卒，需要走到目标 $B$ 点。卒行走的规则：可以向下、或者向右。同时在棋盘上 $C$ 点有一个对方的马，该马所在的点和所有跳跃一步可达的点称为对方马的控制点。因此称之为“马拦过河卒”。

棋盘用坐标表示，$A$ 点 $(0, 0)$、$B$ 点 $(n, m)$，同样马的位置坐标是需要给出的。

![](https://cdn.luogu.com.cn/upload/image_hosting/ipmwl52i.png)

现在要求你计算出卒从 $A$ 点（不经过马的控制点）能够到达 $B$ 点的路径的条数，假设马的位置是固定不动的，并不是卒走一步马走一步。

## 输入格式

一行四个正整数，分别表示 $B$ 点坐标和马的坐标。

## 输出格式

一个整数，表示所有的路径条数。

## 输入输出样例 #1

### 输入 #1

```
6 6 3 3

```

### 输出 #1

```
6

```

## 说明/提示

对于 $100 \%$ 的数据，$1 \le n, m \le 20$，$0 \le$ 马的坐标 $\le 20$。

保证起点不是马的控制点。

**【题目来源】**

NOIP 2002 普及组第四题

````python
# 输入：B(n, m) 马(hx, hy)
n, m, hx, hy = map(int, input().split())

# danger[i][j] 表示点 (i, j) 是否是“危险点”
# 这里多开一圈（+2）是为了方便边界判断，避免访问越界
danger = [[False] * (m + 2) for _ in range(n + 2)]

# 马所在位置本身就是危险点
danger[hx][hy] = True

# 8 个马步的相对位移
dx = [-2, -1, 1, 2, 2, 1, -1, -2]
dy = [1, 2, 2, 1, -1, -2, -2, -1]

# 把马的控制点也标记为危险点
for i in range(8):
    nx = hx + dx[i]
    ny = hy + dy[i]
    # 只有在棋盘范围内才算有效控制点
    if 0 <= nx <= n and 0 <= ny <= m:
        danger[nx][ny] = True

# dp[i][j] 表示从起点 (0, 0) 走到 (i, j) 的路径数
dp = [[0] * (m + 1) for _ in range(n + 1)]
dp[0][0] = 1  # 起点只有一种走法：本身

# 初始化第一列：只能从上往下走
for i in range(1, n + 1):
    if danger[i][0]:
        break  # 遇到危险点，下面的点都无法到达
    dp[i][0] = dp[i - 1][0]

# 初始化第一行：只能从左往右走
for j in range(1, m + 1):
    if danger[0][j]:
        break  # 遇到危险点，右边的点都无法到达
    dp[0][j] = dp[0][j - 1]

# 动态规划遍历整个棋盘
for i in range(1, n + 1):
    for j in range(1, m + 1):
        if danger[i][j]:
            dp[i][j] = 0  # 危险点不能走
        else:
            # 只能从上方或左方走来
            dp[i][j] = dp[i - 1][j] + dp[i][j - 1]

print(dp[n][m])
````
