```python
# NumPy 代码块 1：创建数组并查看基本属性
import numpy as np

array = np.array([1, 2, 3, 4, 5])
zeros = np.zeros((2, 3))
ones = np.ones((2, 2))
sequence = np.arange(0, 10, 2)
print("array:", array)
print("zeros:\n", zeros)
print("ones:\n", ones)
print("arange:", sequence)
print("shape:", array.shape, "ndim:", array.ndim, "size:", array.size, "dtype:", array.dtype)
```

```python
# NumPy 代码块 2：索引、切片和广播
import numpy as np

matrix = np.arange(12).reshape(3, 4)
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])
print(matrix[1, 2], matrix[:, 1], matrix[matrix > 7])
print(x + y)
print(np.arange(3).reshape(3, 1) + np.arange(3))
```

```python
# NumPy 代码块 3：统计分析和线性代数
import numpy as np

x = np.array([1, 2, 3])
matrix = np.array([[1, 2], [3, 4]])
print("统计:", np.max(x), np.mean(x), np.std(x), np.argmax(x))
print("逆矩阵:\n", np.linalg.inv(matrix))
print("行列式:", np.linalg.det(matrix))
```

## 五、NumPy 数组运算实验

### 5.1 实验原理

NumPy 的核心对象是多维数组 `ndarray`。数组中的元素通常具有相同的数据类型，可以通过统一的向量化运算完成批量计算。与逐个处理元素相比，向量化运算代码更简洁，也更适合进行科学计算。

本部分依次完成数组创建、形状查看、索引切片、布尔筛选、广播运算、统计分析、矩阵求逆、行列式计算以及数组合并与拆分。

### 5.2 结果分析

从运行结果可以看出，`shape`、`ndim`、`size` 和 `dtype` 能够快速描述数组的结构；布尔索引可以筛选满足条件的元素；广播机制可以使形状兼容的数组直接参与运算。对二维矩阵计算逆矩阵和行列式，则能够完成基础的线性代数分析。


```python
# Pandas 代码块 1：创建 Series 和 DataFrame
import pandas as pd
import numpy as np

series = pd.Series([10, 20, 30], index=["a", "b", "c"], name="销量")
dataframe = pd.DataFrame({
    "姓名": ["张三", "李四", "王五"],
    "成绩": [88, 92, 75],
    "班级": ["1班", "1班", "2班"],
})
print(series)
print("\n", dataframe)
```

```python
# Pandas 代码块 2：选择、增加和清洗
import pandas as pd
import numpy as np

dataframe = pd.DataFrame({"姓名": ["张三", "李四", "王五"], "成绩": [88, 92, 75]})
dataframe["优秀"] = dataframe["成绩"] >= 85
missing = pd.DataFrame({"A": [1, np.nan, 3], "B": [4, 5, np.nan]})
print(dataframe)
print("缺失值:\n", missing.isna().sum())
print("填充后:\n", missing.fillna(0))
```

```python
# Pandas 代码块 3：分组、排序、合并和透视
import pandas as pd

scores = pd.DataFrame({"性别": ["男", "女", "男", "女"], "成绩": [88, 92, 75, 95]})
print("平均分:\n", scores.groupby("性别")["成绩"].mean())
print("排序:\n", scores.sort_values("成绩", ascending=False))
print("透视:\n", scores.pivot_table(index="性别", values="成绩", aggfunc="mean"))
```

```python
# Pandas 代码块 4：时间序列和文件读写
import pandas as pd

sales = pd.DataFrame({
    "日期": pd.date_range("2026-01-01", periods=5),
    "销量": [12, 15, 14, 20, 18],
})
print(sales.resample("ME", on="日期")["销量"].sum())
sales.to_csv("demo_df.csv", index=False, encoding="utf-8-sig")
print("CSV 已保存")
```

### 7.1 实验原理

Matplotlib 是 Python 中常用的数据可视化库。不同图表适合表达不同类型的信息：折线图用于观察趋势，散点图用于观察变量关系，柱状图用于比较类别，直方图用于观察数据分布，饼图用于展示比例，箱线图用于比较分布特征。

本部分还练习了图例、标题、数值标注、网格线和注释等图表修饰方法，使图形更容易阅读和解释。

### 7.2 结果分析

绘制结果能够直观展示正弦和余弦曲线、随机数据的相关趋势、不同科目的成绩差异以及随机样本的分布情况。箱线图可以同时反映各组数据的中位数、离散程度和异常值，曲线标注则可以突出数据中的最大值位置。


```python
# Matplotlib 代码块 1：折线图和散点图
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
from IPython.display import display, Image

rng = np.random.default_rng(42)
figure, axes = plt.subplots(1, 2, figsize=(10, 4))
x = np.linspace(0, 2 * np.pi, 100)
axes[0].plot(x, np.sin(x), label="sin(x)")
axes[0].plot(x, np.cos(x), "--", label="cos(x)")
axes[0].set_title("折线图")
axes[0].legend()
xs = rng.normal(0, 1, 100)
ys = 0.8 * xs + rng.normal(0, 0.5, 100)
axes[1].scatter(xs, ys, c=ys, cmap="viridis")
axes[1].set_title("散点图")
figure.tight_layout()
figure.savefig("plot_line_scatter.png", dpi=150)
display(Image(filename="plot_line_scatter.png"))
```

```python
# Matplotlib 代码块 2：柱状图和直方图
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Image

rng = np.random.default_rng(42)
figure, axes = plt.subplots(1, 2, figsize=(8, 3))
subjects = ["Python", "数学", "英语", "物理"]
bars = axes[0].bar(subjects, [88, 92, 75, 81], color="tab:orange")
axes[0].bar_label(bars, fmt="%d")
axes[0].set_title("柱状图")
axes[1].hist(rng.normal(100, 15, 1000), bins=30, color="tab:green")
axes[1].set_title("直方图")
figure.tight_layout()
figure.savefig("plot_bar_hist.png", dpi=150)
display(Image(filename="plot_bar_hist.png"))
```

```python
# Matplotlib 代码块 3：饼图和箱线图
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Image

rng = np.random.default_rng(42)
figure, axes = plt.subplots(1, 2, figsize=(8, 3))
axes[0].pie([35, 25, 20, 20], labels=["A", "B", "C", "D"], autopct="%.1f%%")
axes[0].set_title("饼图")
groups = [rng.normal(mean, 5, 60) for mean in (60, 65, 70, 75)]
axes[1].boxplot(groups, tick_labels=["甲", "乙", "丙", "丁"], showmeans=True)
axes[1].set_title("箱线图")
figure.tight_layout()
figure.savefig("plot_pie_box.png", dpi=150)
display(Image(filename="plot_pie_box.png"))
```

```python
# Matplotlib 代码块 4：标注和网格
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Image

x = np.linspace(0, 10, 200)
y = np.exp(-x / 5) * np.cos(x)
figure, axis = plt.subplots(figsize=(7, 3))
axis.plot(x, y, label="衰减曲线")
peak = np.argmax(y)
axis.annotate("最大值", (x[peak], y[peak]), xytext=(x[peak] + 1, 0.7), arrowprops={"arrowstyle": "->"})
axis.grid(alpha=0.3)
axis.legend()
axis.set_title("带标注的曲线")
figure.tight_layout()
figure.savefig("plot_annotate.png", dpi=150)
display(Image(filename="plot_annotate.png"))
```

## 不规则子图布局

```python
# 创建一个 7×5 英寸的绘图画布
df = pd.DataFrame()

fig = plt.figure(figsize=(7, 5))

# 使用 subplot2grid 构建不规则子图布局
# 布局：2 行 3 列
# 左侧子图占据两行，其中（0, 0）位置是左上角
ax1 = plt.subplot2grid((2, 3), (0, 0), rowspan=2)
# 中上方子图占据一个网格，（0, 1）位置是中上方
ax2 = plt.subplot2grid((2, 3), (0, 1))
# 中下方子图占据一个网格,（1, 1）位置是中下方
ax3 = plt.subplot2grid((2, 3), (1, 1))
# 右侧子图占据两行，（0, 2）位置是右上角
ax4 = plt.subplot2grid((2, 3), (0, 2), rowspan=2)
```

```python
# 创建一个空的 DataFrame，后续可用于存放数据
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 创建一个 7×5 英寸的绘图画布
df = pd.DataFrame()

fig = plt.figure(figsize=(7, 5))

# 使用 subplot2grid 构建不规则子图布局
# 布局：2 行 3 列
# 左侧子图占据两行，其中（0, 0）位置是左上角
ax1 = plt.subplot2grid((3, 2), (0, 0), colspan=2)
# 中上方子图占据一个网格，（0, 1）位置是中上方
ax2 = plt.subplot2grid((3, 2), (1,0 ))
# 中下方子图占据一个网格,（1, 1）位置是中下方
ax3 = plt.subplot2grid((3, 2), (1, 1))
# 右侧子图占据两行，（0, 2）位置是右上角
ax4 = plt.subplot2grid((3, 2), (2, 0), colspan=2)
```

### 代码核心解析

1. **网格定义 `(2,3)`**：在画布上划分虚拟网格，一共2行、3列，总共6个单元格。行、列索引从0开始。
2. **ax1**：起始位置`(0,0)`，`rowspan=2`，纵向跨越2行，合并上下两个单元格，形成左侧通高的子图。
3. **ax2**：起始位置`(0,1)`，默认仅占用1个单元格，对应网格第0行第1列。
4. **ax3**：起始位置`(1,1)`，仅占用1个单元格，对应网格第1行第1列。
5. **ax4**：起始位置`(0,2)`，`rowspan=2`，纵向跨越2行，合并上下单元格，形成右侧通高的子图。

### 关键知识点
`subplot2grid` 用于实现**非等尺寸子图布局**。
- 普通`subplot()`只能均分画布，所有子图大小一致；
- `subplot2grid`基于虚拟网格，通过`rowspan`（纵向跨行）、`colspan`（横向跨列）实现单元格合并，灵活构造不规则排版。
